```python
#Step 6: Time-based analysis 
type(df["Order Date"]) 
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Order Date"]
```

```python
sales_time = df.groupby("Order Date")["Sales"].sum()
x = sales_time.index
y = sales_time.values
fig, ax = plt.subplots()
ax.plot(y,x)
ax.set(xlabel='Total Sales', ylabel='Month-Year',
       title='Month-Year trend of total sales')
plt.show()
```

```python
sbn.lineplot(x=y,y=x)
plt.title("Month-Year trend of total sales")
plt.xlabel("Month-Year")
plt.ylabel("Total Sales")
plt.show()
```

```python
#Insights
#Sales show an overall increasing trend from 2015 to 2018, indicating business growth over time. 
#However, the pattern is not consistent and exhibits clear fluctuations across months. 
#There is strong seasonality, with sales peaking consistently in the later months of the year, particularly from September to December. 
#In contrast, the early months of the year, especially January and February, show relatively lower sales. 
#This suggests that revenue is influenced by seasonal demand, with significant dependence on year-end performance.
```

```python
#Overall insights
#The overall sales performance shows a clear upward trend from 2015 to 2018, indicating steady business growth over time. 

#Technology emerges as the leading category, primarily driven by high-performing sub-categories such as Phones, suggesting that revenue is concentrated in a few key products. In contrast, Furniture underperforms, with lower contribution from its sub-categories.

#From a customer perspective, the Consumer segment contributes the highest share of revenue, indicating strong demand from individual buyers. The distribution is skewed, with relatively lower contributions from Corporate and Home Office segments.

#Regionally, the West region generates the highest sales, while the South region contributes the least, highlighting geographical disparities in performance.

#Time-based analysis reveals strong seasonality, with sales consistently peaking towards the end of the year, particularly between September and December, while the early months show lower performance. This indicates a reliance on year-end demand.

#Overall, the business shows growth but is dependent on a limited set of products, customer segments, and seasonal demand patterns. This suggests opportunities to diversify revenue sources across underperforming categories, segments, and regions.
```
