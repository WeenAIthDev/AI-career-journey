```python
#Step 2: Which region generates more sales for SuperStore
#Dimension: Region
#Metrics: Sales 
#So we have to investigate sales by region
#region = df.groupby("Region") # groups rows with respect to category Region percent = (region_sales/region_sales.sum())*100 #region sales refers to the total sales series its a series
region_sales = df.groupby("Region")["Sales"].sum() # Add up sales based on region
region_sales
```

```python
#Inference 
#Which region leads sales?
#region_sales.loc[region_sales == region_sales.max()]
region_sales.idxmax()
```

```python
#Which region is weakest?
#region_sales.loc[region_sales == region_sales.min()]
region_sales.idxmin()
```

```python
#Is the distribution balanced?
percent = (region_sales/region_sales.sum())*100 #region sales refers to the total sales series its a series
percent
```

```python
#Observation:
#West (~31%) and East (~29%) contribute the majority of sales.

#Explanation:
#Sales are concentrated in fewer regions, indicating uneven distribution.

#Business implication:
#The company should investigate underperforming regions like South to identify growth opportunities.
```

```python
# Visual Confirmation

# using seaborn 
sbn.barplot(x=region_sales.index, y=region_sales.values)
plt.title("Total Sales By Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
# using matplotlib

fig, ax = plt.subplots()# briinging a chart fig and placing scale on it the axes thru ax
ax.bar(["Central", "East", "South", "West"], percent, color=["tab:red", "tab:blue", "tab:orange", "tab:green"], label=["Central", "East", "South", "West"])
ax.set_title("Percentage Of Sales")
ax.set_ylabel("Sales percent")
ax.set_xlabel("Regions of Sales")
ax.legend(title="Regions in US")
plt.show()
```

```python
#Step 6: Time-based analysis 
type(df["Order Date"]) 
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Order Date"]
```

```python
datetime_year = df["Order Date"].dt.year
datetime_year
```

```python
datetime_month = df["Order Date"].dt.month
datetime_month
```

```python
sales_time = df.groupby([datetime_month])["Sales"].sum()
sales_time
```

```python
#Is sales increasing over time?
#Is it fluctuating?
#Are there peaks and drops?
```

```python
x = sales_time.index
y = sales_time.values
fig, ax = plt.subplots()
ax.plot(x,y)
ax.set(xlabel='Month', ylabel='Total Sales',
       title='Monthly trend of total sales')
plt.show()
```

```python
sbn.lineplot(x=x,y=y)
plt.title("Monthly trend of total sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Insights
#Sales show an overall increasing trend from 2015 to 2018, indicating business growth over time. 
#However, the pattern is not consistent and exhibits clear fluctuations across months. 
#There is strong seasonality, with sales peaking consistently in the later months of the year, particularly from September to December. 
#In contrast, the early months of the year, especially January and February, show relatively lower sales. 
#This suggests that revenue is influenced by seasonal demand, with significant dependence on year-end performance.
```
